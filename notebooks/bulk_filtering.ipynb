```python
# Create a list to store alloys that contain specific elements
alloy_data = []

# Define the list of elements to filter alloys
Atom_list_primary = ['Fe', 'Co', 'Ni', 'Cu', 'Ru', 'Rh', 'Pd', 'Ag', 'Os', 'Ir', 'Pt', 'Au']
Atom_list_secondary = ['Cr', 'Zn', 'Y', 'Zr', 'In', 'Sn', 'Pb', 'Bi']
Atom_list = Atom_list_primary + Atom_list_secondary

# Filter alloys to include only those with specific elements
for alloy in unique_alloy_list:
    if all(element in Atom_list for element in alloy.get_chemical_symbols()):
        alloy_data.append(alloy)

print(len(alloy_data))
for i in range(len(Atom_list)):
    print(Atom_list[i], end = ", ")
```

```python
alloy_data = []

# Define the list of elements to filter alloys
Atom_list_primary = ['Fe', 'Co', 'Ni', 'Cu', 'Ru', 'Rh', 'Pd', 'Ag', 'Os', 'Ir', 'Pt', 'Au']
Atom_list_secondary = ['Cr', 'Zn', 'Y', 'Zr', 'In', 'Sn', 'Pb', 'Bi']

i = 0

for alloy in unique_alloy_list:
    elements = set(alloy.get_chemical_symbols())
    
    # Only keep bimetallic alloys
    if len(elements) != 2 and len(elements) != 1:
        print(f'This element is not bimetallic or an urnary metal: {alloy}')
        i += 1
        continue

    # Rule: at least one element in primary
    if not any(el in Atom_list_primary for el in elements):
        print(f'This alloy does not contain a valid primary element: {alloy.symbols}')
        i+=1
        continue
    
    # Rule: all elements must be in primary or secondary
    if not all(el in Atom_list_primary + Atom_list_secondary for el in elements):
        print(f'This alloy does not contain a valid secondary element: {alloy.symbols}')
        i+=1
        continue
    
    alloy_data.append(alloy)

print(f'{i} alloys were unfit for screening, the number of valid alloys is: {len(alloy_data)}')
```

```python
missing = [str(i.symbols) for i in copper_alloys if i not in alloy_data]
missing_filtered = [i for i in missing if not any(sub in i for sub in ["Ca", "Mg", "Sr", "Se", "Ba", "Po", "At", "Tc", "Rb", ])]
print(missing_filtered, len(missing_filtered))
```
