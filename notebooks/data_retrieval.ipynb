```python
import pickle
import numpy as np
from ocdata.core import Bulk


Element_List = ['Pt', 'Ru', 'Ir', 'Ni', 'Cu', 'Pd', 'Co', 'Rh', 'Fe', 'Ag', 'Au']

Element_List = ['V', 'Cr', 'Mn', 'Fe', 'Co', 'Ni', 'Cu', 'Zn', 'Ru', 'Rh', 'Pd', 'Ag', 'Ir', 'Pt', 'Au']

bulk_db_path = 'pkls/bulks.pkl'

bulk_db = pickle.load(open(bulk_db_path, "rb"))

```

```python
bulk = Bulk(bulk_id_from_db=0, bulk_db_path=bulk_db_path)
bulk_ele = list(set(bulk.atoms.get_chemical_symbols()))
print(bulk_ele)
select = True
for _ele in bulk_ele:
    if _ele not in Element_List:
        select = False
        break
```

```python
Element_List = ['Pt', 'Ru', 'Ir', 'Ni', 'Cu', 'Pd', 'Co', 'Rh', 'Fe', 'Ag', 'Au']
Element_List = [78, 44, 77, 28, 29, 46, 27, 45, 26, 47, 79]

bulk_list = []
for i in range(len(bulk_db)):
    # bulk = Bulk(bulk_id_from_db=i, bulk_db_path=bulk_db_path)
    # print(i)
    bulk_obj = bulk_db[i]
    atoms, src_id = bulk_obj["atoms"], bulk_obj["src_id"]
    bulk_ele = np.unique(atoms.get_atomic_numbers())
    select = True
    for _ele in bulk_ele:
        if _ele not in Element_List:
            select = False
            break
    if select:
        bulk_list.append(atoms)


print(len(bulk_list))
with open('select_bulks2.pkl', 'wb') as f:
    pickle.dump(bulk_list, f)
```
